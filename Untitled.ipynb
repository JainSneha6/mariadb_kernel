```sql
CREATE TABLE employees (
    emp_id INT PRIMARY KEY AUTO_INCREMENT,
    name VARCHAR(50),
    department VARCHAR(50),
    age INT,
    gender VARCHAR(10),
    education_level VARCHAR(30),
    years_experience INT,
    projects_completed INT,
    avg_project_score DECIMAL(5,2),
    certifications INT,
    training_hours INT,
    overtime_hours INT,
    remote_ratio DECIMAL(3,2),
    salary DECIMAL(10,2),
    bonus DECIMAL(10,2),
    satisfaction_score DECIMAL(3,2),
    performance_rating INT,        -- target variable for classification
    potential_score DECIMAL(5,2),  -- target variable for regression
    attrition_flag INT             -- 1 = left company, 0 = stayed
);
```
